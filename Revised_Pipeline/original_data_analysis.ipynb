```python
human_gr = pd.read_csv(
    r"F:\biotecnika\data stored\originalGSE108966\Original_Human.GRCh38.p13.annot.tsv",
    sep="\t",
    comment="#",
    dtype=str,
    low_memory=False
)

human_gr.head()
```

```python
# Show all column names
print(human_gr.columns.tolist())

mirnas = ["3607", "5701", "4510", "378g"]

# Search columns containing symbol, synonym, name, alias, or identifier
search_columns = [
    column for column in human_gr.columns
    if any(
        word in column.lower()
        for word in ["symbol", "synonym", "name", "alias", "id"]
    )
]

print("Columns being searched:")
print(search_columns)

if not search_columns:
    print("No relevant columns were found.")

else:
    # Match different formats:
    # miR-3607
    # miR3607
    # MIR-3607
    # hsa-miR-3607
    # hsa_miR_3607
    pattern = r"(?i)(?:hsa[-_ ]?)?mir[-_ ]?(?:3607|5701|4510)(?!\d)"

    # Search all selected columns
    matches = (
        human_gr[search_columns]
        .fillna("")
        .astype(str)
        .apply(
            lambda col: col.str.contains(
                pattern,
                regex=True,
                na=False
            )
        )
        .any(axis=1)
    )

    # Return the COMPLETE rows
    mirna_matches = human_gr.loc[matches].copy()

    print("Number of matches:", len(mirna_matches))

    display(mirna_matches)
```

```python
gse_id = "GSE108966"
import GEOparse

gse = GEOparse.get_GEO(
    geo=gse_id,
    destdir="."
)
sample_data = []

for gsm_name, gsm in gse.gsms.items():
    data = gsm.metadata
    
    sample_data.append({
        "Accession": gsm_name,
        "Title": data.get("title", [""])[0],
        "Source name": data.get("source_name_ch1", [""])[0]
    })

metadata = pd.DataFrame(sample_data)

display(metadata)
```

```python
# Normalize the source-name values
metadata["Source name"] = metadata["Source name"].str.strip().str.lower()

# Separate samples
endometrium_metadata = metadata[
    metadata["Source name"].eq("endometrium")
].reset_index(drop=True)

blood_metadata = metadata[
    metadata["Source name"].eq("blood")
].reset_index(drop=True)

# Display both tables
display(endometrium_metadata)
display(blood_metadata)


print("Endometrium samples:", len(endometrium_metadata))
print("Blood samples:", len(blood_metadata))
```
